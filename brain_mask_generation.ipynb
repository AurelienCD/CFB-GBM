## Generate brain mask

```python
import subprocess
from pathlib import Path
from tqdm import tqdm

# Define your root directories
INPUT_DIR = Path("./CFB-GBM/data")
TARGET_DIR = Path("./CFB-GBM_brain_mask/data")

# Supported file extensions
IMAGE_EXTENSIONS = (".nii", ".nii.gz")

print("Starting batch brain extraction...")
# 1. Loop through all files in the input directory recursively
for mri_t1gd in tqdm(INPUT_DIR.rglob("*t1gd.nii.gz")):
    patient = mri_t1gd.name.split('_')[0]
    temporality = mri_t1gd.name.split('_')[1]
    target_file_dir = TARGET_DIR / patient / temporality
    target_file_dir.parent.mkdir(parents=True, exist_ok=True)

    mask_output_file = target_file_dir / f"{patient}_{temporality}_brain-mask.nii.gz"

    #
    # print(f"\n--- Processing: {mri_t1gd.parent.name} ---")
    # print(f"Input:  {mri_t1gd}")
    # print(f"Mask Output: {mask_output_file}")

    # 5. Construct the shell command as a list of strings (safer than a single string)
    # Using HD-BET as the example tool here
    command = [
        "hd-bet",
        "-i", str(mri_t1gd),
        # "-o", str(target_file_dir / "t1gd_clean.nii.gz"), # Temp clean output required by HD-BET
        "-o", str(mask_output_file),
        "--save_bet_mask"
    ]

    try:
        # 6. Execute the shell command safely inside Python
        # text=True and capture_output=True allows you to catch errors or logs if needed
        result = subprocess.run(command, check=True, text=True, capture_output=True)
        # print(f"Successfully generated mask for {mri_t1gd.parent.name}")

        # Optional: If you only wanted the mask and not the cleaned T1gd generated by HD-BET,
        # you can delete the temporary clean file to save space:
        # (target_file.parent / "t1gd_clean.nii.gz").unlink(missing_ok=True)

    except subprocess.CalledProcessError as e:
        # print(f"ERROR processing {mri_t1gd.name}:")
        print(e.stderr)
        # Depending on your preference, you can 'continue' to next patient or 'raise' to stop the loop
        continue

print("\nAll patients processed successfully!")
```

## Rename with the right format

```python
source_dir = Path("./CFB-GBM_brain_mask/data")
for processed_mri in tqdm(source_dir.rglob("*brain-mask.nii.gz")):
    if processed_mri.is_file():
        print(f'deleting {processed_mri.name}')
        processed_mri.unlink() # delete the mri
for mask_mri in tqdm(source_dir.rglob("*brain-mask_bet.nii.gz")):
    patient = mask_mri.name.split('_')[0]
    temporality = mask_mri.name.split('_')[1]
    new_file = mask_mri.parent / f"{patient}_{temporality}_brain_mask.nii.gz"
    print(f'rename {mask_mri.name} to {new_file.name}')
    mask_mri.rename(new_file)
```
